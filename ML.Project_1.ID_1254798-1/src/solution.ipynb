### iii. Разница между multiclass и multilabel классификацией

#### Multiclass Classification (Многоклассовая классификация)

**Определение:** Каждый объект принадлежит **ровно одному** классу из нескольких возможных (количество классов ≥ 3).

**Примеры:**
- Распознавание цифр 0-9: изображение содержит ровно одну цифру
- Классификация типа животного: кошка, собака, птица (объект может быть только чем-то одним)
- Определение основного жанра фильма: комедия, драма, боевик, триллер (выбираем один основной)
- Классификация цветка по виду: ирис, роза, тюльпан


#### Multilabel Classification (Мультиметочная классификация)

**Определение:** Каждый объект может одновременно принадлежать **нескольким классам** (или ни одному).

**Примеры:**
- Теги для статьи: может быть одновременно "технологии", "наука", "бизнес", "образование"
- Жанры фильма: может быть одновременно комедия, боевик, приключения
- Симптомы заболевания: пациент может одновременно иметь головную боль, тошноту, лихорадку
- Объекты на фотографии: на одном изображении могут присутствовать кошка, собака, человек, дерево

### iv. Задача с ценами на дома: классификация или регрессия?

#### Ответ: Это задача **регрессии**

**Обоснование:**

Задача предсказания цены дома является типичной задачей регрессии по следующим причинам:

1. **Целевая переменная непрерывна:** Цена дома — это непрерывная числовая величина, которая может принимать любое значение в определенном диапазоне (например, $150,000.00, $250,350.50, $1,234,567.89)

2. **Бесконечное множество возможных значений:** В отличие от классификации, где количество классов конечно, цена может быть любым числом

3. **Важна точность предсказания:** Нам нужно предсказать конкретную числовую величину, а не просто отнести дом к категории

#### Можно ли свести регрессию к классификации?

**Да, технически это возможно.** Существует несколько подходов:

##### Пример: Дискретизация (Binning)

Разбить непрерывный диапазон цен на дискретные интервалы (bins):
```
Класс 1: "Эконом" — $0 - $100,000
Класс 2: "Стандарт" — $100,000 - $200,000
Класс 3: "Комфорт" — $200,000 - $300,000
Класс 4: "Премиум" — $300,000 - $500,000
Класс 5: "Люкс" — $500,000+
```

**Преимущества:**
- Простота интерпретации результатов
- Устойчивость к выбросам
- Может быть полезно для бизнес-задач (сегментация рынка)

**Недостатки:**
- **Потеря точности:** Дом за $199,999 и дом за $200,001 попадут в разные классы, хотя их цены практически идентичны
- **Произвольность границ:** Выбор границ интервалов субъективен и может сильно влиять на результат
- **Потеря информации:** Внутри класса теряется информация о точном значении
- **Не учитывается порядок:** Стандартные алгоритмы классификации не учитывают, что класс 2 ближе к классу 3, чем к классу 1

c. Drop the rows that are outside the 1 and 99 percentiles from the target column.

````python
def remove_outliers(dataframe: pd.DataFrame, cols: list[str], low: int = 1, high: int = 99) -> pd.DataFrame:
  for col in cols:
    P_low = np.percentile(df[col], low)
    P_high = np.percentile(df[col], high)
    dataframe = dataframe[(dataframe[col] > P_low) & (dataframe[col] < P_high)]
  return dataframe.reset_index(drop=True)
````

````python
df = remove_outliers(df, ["price"], low=1)
df
````
