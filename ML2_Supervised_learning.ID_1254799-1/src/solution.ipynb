### 1. Derive an analytical solution to the regression problem. Use a vector form of the equation.

Аналитическое решение находят путем минимизации функции потерь MSE (Mean Squared Error).

#### 1.1. Функция потерь в векторной форме (MSE)

Функция потерь $J(\mathbf{w})$ для вектора весов $\mathbf{w}$ имеет вид:

$$
J(\mathbf{w}) = \frac{1}{2m} ||\mathbf{X}\mathbf{w} - \mathbf{y}||^2
$$

Где:
* $\mathbf{w}$ — вектор весов (параметров), который мы ищем.
* $\mathbf{X}$ — матрица признаков (с добавленным столбцом единиц для смещения/intercept).
* $\mathbf{y}$ — вектор целевых значений.
* $m$ — количество обучающих примеров.

#### 1.2. Вывод нормального уравнения

Чтобы найти минимум, необходимо взять частную производную функции потерь по вектору весов $\mathbf{w}$ и приравнять её к нулю ($\nabla_{\mathbf{w}} J(\mathbf{w}) = 0$).

1.  Раскрываем норму и переписываем функцию потерь:
    $$
    J(\mathbf{w}) = \frac{1}{2m} (\mathbf{X}\mathbf{w} - \mathbf{y})^T (\mathbf{X}\mathbf{w} - \mathbf{y})
    $$

2.  Берём градиент:
    $$
    \nabla_{\mathbf{w}} J(\mathbf{w}) = \frac{1}{m} \mathbf{X}^T (\mathbf{X}\mathbf{w} - \mathbf{y})
    $$

3.  Приравниваем градиент к нулю:
    $$
    \mathbf{X}^T (\mathbf{X}\mathbf{w} - \mathbf{y}) = 0
    $$

4.  Выражаем вектор весов $\mathbf{w}$:
    $$
    \mathbf{X}^T \mathbf{X} \mathbf{w} - \mathbf{X}^T \mathbf{y} = 0
    $$
    $$
    \mathbf{X}^T \mathbf{X} \mathbf{w} = \mathbf{X}^T \mathbf{y}
    $$
    $$
    \mathbf{w} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}
    $$

#### 1.3. Аналитическое решение

Вектор весов $\mathbf{w}$, который минимизирует MSE, определяется формулой:
$$\mathbf{w} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}$$

Это и есть **нормальное уравнение**.

### 2. What changes in the solution when L1 and L2 regularizations are added to the loss function.

Регуляризация добавляет к основной функции потерь MSE член, штрафующий большие значения весов $\mathbf{w}$.

#### 2.1. L2 Регуляризация (Ridge Regression)

* **Функция потерь:** $J_{\text{Ridge}}(\mathbf{w}) = J(\mathbf{w}) + \alpha ||\mathbf{w}||_2^2$
    $$
    J_{\text{Ridge}}(\mathbf{w}) = \frac{1}{2m} ||\mathbf{X}\mathbf{w} - \mathbf{y}||^2 + \alpha \sum_{j=1}^{n} w_j^2
    $$
    Где $\alpha \ge 0$ — коэффициент регуляризации.

* **Аналитическое решение (Нормальное уравнение для Ridge):**
    Добавление L2-штрафа делает матрицу $(\mathbf{X}^T \mathbf{X})$ всегда обратимой (даже при мультиколлинеарности), стабилизируя решение.
    $$\mathbf{w}_{\text{Ridge}} = (\mathbf{X}^T \mathbf{X} + \alpha \mathbf{I})^{-1} \mathbf{X}^T \mathbf{y}$$
    Где $\mathbf{I}$ — единичная матрица, и $\mathbf{I}$ имеет размерность $n \times n$ (где $n$ — количество признаков).

#### 2.2. L1 Регуляризация (Lasso Regression)

* **Функция потерь:** $J_{\text{Lasso}}(\mathbf{w}) = J(\mathbf{w}) + \alpha ||\mathbf{w}||_1$
    $$
    J_{\text{Lasso}}(\mathbf{w}) = \frac{1}{2m} ||\mathbf{X}\mathbf{w} - \mathbf{y}||^2 + \alpha \sum_{j=1}^{n} |w_j|
    $$

* **Аналитическое решение (Lasso):**
    Из-за наличия модуля ($|w_j|$) функция потерь для Lasso **недифференцируема в нуле**, то есть у неё **не существует замкнутого аналитического решения** (аналогичного нормальному уравнению).

    Поэтому Lasso-регрессия решается **итерационными методами**, такими как:
    1.  **Координатный спуск (Coordinate Descent).**
    2.  **Субградиентный спуск (Subgradient Descent)** или, чаще, модификации Градиентного спуска (например, FISTA).

### 3. Explain why L1 regularization is often used to select features. Why are there many weights equal to 0 after the model is fit?

L1-регуляризация (Lasso) используется для выбора признаков, поскольку она приводит к тому, что веса **многих неинформативных признаков становятся точно равны нулю**. 

#### 3.1. Геометрическая причина (Контурные линии)

Это свойство можно объяснить, рассмотрев геометрическое представление минимизации. Задача минимизации заключается в поиске точки касания между:
1.  **Эллиптическими контурами** функции потерь MSE.
2.  **Областью ограничений** регуляризатора.

* **L2 (Ridge):** Область ограничений — это **круг** (или сфера). Точка касания с эллипсом MSE почти всегда происходит **не на осях**, а где-то в пространстве. Следовательно, веса $w_j$ становятся **близки к нулю**, но редко *равны* нулю.

* **L1 (Lasso):** Область ограничений — это **ромб** (или многогранник с "острыми углами" на осях). Когда контурная линия MSE касается ромба, она **очень часто касается его в одном из этих углов**. Точка касания, находящаяся на оси, соответствует ситуации, когда вес, связанный с этой осью, **обнуляется** (становится $w_j = 0$).

#### 3.2. Почему веса становятся 0

Математически, штраф $L1$ ($\sum |w_j|$) имеет тенденцию "выталкивать" веса к нулю, если они не вносят существенного вклада в снижение ошибки.

Когда модель обучена (fit), веса, равные 0, означают, что **соответствующий признак был исключен из модели**. Таким образом, L1-регуляризация автоматически выполняет **выбор признаков (Feature Selection)**.

### 4. Explain how you can use the same models (Linear regression, Ridge, etc.) but make it possible to fit nonlinear dependencies.

Модели, такие как Линейная регрессия, Ridge и Lasso, по своей природе являются **линейными по параметрам (весам)**, но они могут моделировать **нелинейные зависимости между признаками и таргетом**.

#### 4.1. Метод: Полиномиальная регрессия (Polynomial Features)

Ключевой метод — **создание нелинейных признаков** из существующих **до** обучения модели. Это достигается с помощью **полиномиальных преобразований** и **взаимодействий признаков (Interaction Terms)**.

**Пример:**
Пусть исходная зависимость нелинейна: $y = w_0 + w_1 x_1^2 + w_2 x_2 + \epsilon$

1.  **Трансформация признаков:** Мы создаём новый признак $x_{new} = x_1^2$.
2.  **Новая модель:** Модель становится линейной по новым признакам: $y = w_0 + w_1 x_{new} + w_2 x_2 + \epsilon$.

Этот процесс называется **Полиномиальная регрессия**. 

#### 4.2. Алгоритм

1.  **Создание полиномиальных признаков:** Используем `sklearn.preprocessing.PolynomialFeatures` для создания признаков высокой степени ($x^2, x^3, \dots$) и признаков взаимодействия ($x_1 x_2, x_1^2 x_2, \dots$) из исходной матрицы $\mathbf{X}$.
2.  **Нормализация:** Обязательно применяем **StandardScaler** к новым, расширенным признакам, чтобы стабилизировать обучение.
3.  **Обучение:** Обучаем линейную модель (Linear Regression, Ridge, Lasso) на **трансформированных признаках**.

Таким образом, модель остается "линейной" в пространстве своих параметров $\mathbf{w}$, но благодаря преобразованию входных данных может аппроксимировать сколь угодно сложную **нелинейную кривую** в исходном пространстве признаков.

### 1. Implement a Python class for a linear regression algorithm with two basic methods — fit and predict. Use stochastic gradient descent (SGD) to find optimal model weights. For better understanding, we recommend implementing separate versions of the algorithm with the analytical solution and non-stochastic gradient descent under the hood.

```python
class LinearRegressionCustom:
  def __init__(self, learning_rate=0.01, n_epochs=1000):
    self.learning_rate = learning_rate
    self.n_epochs = n_epochs
    self.weights = None

  def _add_intercept(self, X):
    """
    Добавляет столбец единиц к матрице X.
    Это нужно, чтобы учесть свободный член (bias) в векторе весов.
    """
    intercept = np.ones((X.shape[0], 1))
    return np.concatenate((intercept, X), axis=1)

  def predict(self, X):
    """
    формула предсказания: y = X * w
    """
    if self.weights is None:
      raise Exception("Model is not trained yet")
    
    print(f"weights={self.weights}")
    X_b = self._add_intercept(X)
    return X_b.dot(self.weights) # скалярное произведение в numpy
  
  # Стохастический градиентный спуск (SGD)
  def fit_sgd(self, X, y):

    if isinstance(X, (np.generic, np.ndarray)):
        X_np = X
    else:
        X_np = X.values # Извлекаем значения из DataFrame
        
    if isinstance(y, (np.generic, np.ndarray)):
        y_np = y
    else:
        y_np = y.values.flatten() # Извлекаем значения из Series и делаем плоским
    
    X_b = self._add_intercept(X_np)
    m = X_np.shape[0]
    n = X_np.shape[1]
    print(f"m={m}, n={n}")

    self.weights = np.random.randn(n + 1) # +1 bias col

    for epoch in range(self.n_epochs):
      # Для SGD важно перемешивать данные каждую эпоху,
      # чтобы циклы обучения не "привыкали" к порядку данных
      indices = np.random.permutation(m)
      X_b_shuffled = X_b[indices]
      y_shuffled = y_np[indices]

      for i in range(m):
        # берем ОДИН случайый пример
        xi = X_b_shuffled[i:i+1]
        yi = y_shuffled[i:i+1]

        # 1. Предсказание для одного примера
        prediction = xi.dot(self.weights)

        # 2. Ошибка для одного примера
        error = prediction - yi

        # 3. Градиент для одного примера
        # (xi.T имеет размерность (n,1), error — скаляр)
        gradient = 2 * xi.T.dot(error)

        # 4. Обновление весов
        # В SGD часто уменьшают learning_rate со временем, 
        # но для простоты оставим его фиксированным
        self.weights = self.weights - self.learning_rate * gradient.flatten()
```

### 3. Define the R squared (R2) coefficient and implement a function to calculate it.

1. Определние коэффициента
R^2 - Коэффициент **детерминации** (R2, R-squared) — это статистическая мера, которая показывает, **насколько хорошо** регрессионная модель (в нашем случае, линейная регрессия) соответствует данным.

R2 показывает, какая доля дисперсии (изменчивости) зависимой переменной (y) объясняется моделью, по сравнению с тем, если бы мы просто использовали среднее значение y.


**Значения R2:**

  - R2=1.0 (или 100%): Модель идеальна. Все предсказанные точки лежат ровно на линии регрессии. Модель идеально объясняет изменчивость данных.

  - R2=0.0 (или 0%): Модель ничего не объясняет. Ее предсказания не лучше, чем использование среднего значения y.

  - R2<0.0: Модель очень плохая. Она хуже, чем использование среднего значения y. Такое может случиться, если модель сильно переобучена или неправильно применена.

**2. Формула R2**
Для расчета R2 нам нужно сравнить три типа дисперсии (суммы квадратов):

  1. Total Sum of Squares (SStotal​): Общая дисперсия данных y. Считает, насколько фактические значения y отклоняются от их среднего (yˉ​).

    SStotal​=i=1∑n​(yi​−yˉ​)2

  2. Residual Sum of Squares (SSres​): Остаточная дисперсия (ошибка). Считает, насколько фактические значения y отклоняются от предсказаний модели (y^​).
  
    SSres​=i=1∑n​(yi​−y^​i​)2

Формула для R2 сравнивает эти две суммы:
R2= 1 - SSres / SStotal​​

Логика: Если ошибка модели (SSres​) мала по сравнению с общей дисперсией (SStotal​), то R2 будет близок к 1.

```python
def r2_score(y_true, y_pred):
  """
  Рассчитывает коэффициент детерминации (R-squared).
    Args:
        y_true (np.ndarray): Фактические значения целевой переменной.
        y_pred (np.ndarray): Предсказанные моделью значения.
        
    Returns:
        float: Значение R^2.
  """

  # 1. Вычисляем среднее значение y_true (y_bar)
  y_mean = np.mean(y_true)

  # 2. Total Sum of Squares (SS_total) - Сумма квадратов отклонений от среднего
  SS_total = np.sum((y_true - y_mean)**2)

  # 3. Residual Sum of Squares (SS_res) - Сумма квадратов ошибок (остатков)
  SS_res = np.sum((y_true - y_pred)**2)

  # 4. Расчет R^2
  R2 = 1 - (SS_res / SS_total)

  return R2
```

### 4. Make predictions with your algorithm and estimate the model with MAE, RMSE and R2 metrics.

```python
def mae_score(y_true, y_pred):
  """
  Рассчитывает Среднюю Абсолютную Ошибку (MAE).
  """
  return np.mean(np.abs(y_true - y_pred))

def rmse_score(y_true, y_pred):
  """
  Рассчитывает Корень из Среднеквадратичной Ошибки (RMSE).
  """
  # Сначала считаем MSE (Mean Squared Error)
  mse = np.mean((y_true - y_pred)**2)
  # Затем берем корень
  return np.sqrt(mse)
```

```python
def create_metric_df(model_name, y_train_true, y_train_pred, y_test_true, y_test_pred):
    """Создает DataFrame для одной модели, собирая R2, MAE и RMSE."""
    
    # Расчет всех метрик
    metrics = {
        'R2': (r2_score(y_train_true, y_train_pred), r2_score(y_test_true, y_test_pred)),
        'MAE': (mae_score(y_train_true, y_train_pred), mae_score(y_test_true, y_test_pred)),
        'RMSE': (rmse_score(y_train_true, y_train_pred), rmse_score(y_test_true, y_test_pred)),
    }
    
    # Создание DataFrame
    df = pd.DataFrame({
        'Model': model_name,
        'Metric': list(metrics.keys()),
        'Train': [m[0] for m in metrics.values()],
        'Test': [m[1] for m in metrics.values()]
    })
    return df
```

## 5. Regularized models implementation — Ridge, Lasso, ElasticNet

### 1. Implement Ridge, Lasso, ElasticNet algorithms: extend the loss function with L2, L1 and both regularizations accordingly.

```python
class LinearRegressionCustom:
  # + параметры регуляризации в __init__
  def __init__(self, learning_rate=0.01, n_epochs=1000, alpha=0.0, l1_ratio=0.0):
    self.learning_rate = learning_rate
    self.n_epochs = n_epochs
    self.weights = None
    # alpha (лямбда) - сила регуляризации
    self.alpha = alpha
    # l1_ratio - соотношения L1/L2 (0.0 - Ridge, 1.0 - Lasso)
    self.l1_ratio = l1_ratio

  def _add_intercept(self, X):
    """
    Добавляет столбец единиц к матрице X.
    Это нужно, чтобы учесть свободный член (bias) в векторе весов.
    """
    intercept = np.ones((X.shape[0], 1))
    return np.concatenate((intercept, X), axis=1)

  def predict(self, X):
    """
    формула предсказания: y = X * w
    """
    if self.weights is None:
      raise Exception("Model is not trained yet")
    
    # print(f"weights={self.weights}")
    X_b = self._add_intercept(X)
    return X_b.dot(self.weights) # скалярное произведение в numpy
  
  # Стохастический градиентный спуск (SGD) + Регуляризация
  def fit_sgd(self, X, y):

    if isinstance(X, (np.generic, np.ndarray)):
        X_np = X
    else:
        X_np = X.values # Извлекаем значения из DataFrame
        
    if isinstance(y, (np.generic, np.ndarray)):
        y_np = y
    else:
        y_np = y.values.flatten() # Извлекаем значения из Series и делаем плоским
    
    X_b = self._add_intercept(X_np)
    m = X_np.shape[0]
    n = X_np.shape[1]

    self.weights = np.random.randn(n + 1) # +1 bias col

    for epoch in range(self.n_epochs):
      # Для SGD важно перемешивать данные каждую эпоху,
      # чтобы циклы обучения не "привыкали" к порядку данных
      indices = np.random.permutation(m)
      X_b_shuffled = X_b[indices]
      y_shuffled = y_np[indices]

      for i in range(m):
        # берем ОДИН случайый пример
        xi = X_b_shuffled[i:i+1]
        yi = y_shuffled[i:i+1]

        # 1. Предсказание для одного примера
        prediction = xi.dot(self.weights)

        # 2. Ошибка для одного примера
        error = prediction - yi

        # 3. Градиент MSE для одного примера
        # (xi.T имеет размерность (n,1), error — скаляр)
        gradient_mse = 2 * xi.T.dot(error).flatten()

        # 4. Градиент регуляризации
        # Создаем вектор штрафа, который не штрафует w0 (bias) - self.weights[0]

        # копируем веса для штрафа, обнуляя w0: [0, w1, w2, ...]
        weights_for_penalty = self.weights.copy()
        weights_for_penalty[0] = 0.0

        # L2 (Ridge) : 2 * alpha * (1 - l1_ratio) * w
        l2_penalty_grad = (2 * self.alpha * (1 - self.l1_ratio)) * weights_for_penalty

        # L1 (Lasso) : alpha * l1_ratio * sign(w)
        l1_penalty_grad = (self.alpha * self.l1_ratio) * np.sign(weights_for_penalty)

        # Полный градиент штрафа
        gradient_penalty = l2_penalty_grad + l1_penalty_grad

        # 5. Обновление весов (штраф добавляется к градиенту)
        full_gradient = gradient_mse + gradient_penalty

    
        # В SGD часто уменьшают learning_rate со временем, 
        # но для простоты оставим его фиксированным
        self.weights = self.weights - self.learning_rate * full_gradient
```

## 6. Feature normalization

### 1. First, write several examples of why and where feature normalization is mandatory and vice versa.

Нормализация признаков (Feature Normalization) необходима, когда алгоритм **чувствителен к масштабу** входных данных, и она может быть необязательна или нежелательна, когда алгоритм **инвариантен к масштабу**. Нормализация обязательна для алгоритмов, основанных на расстоянии или градиентном спуске, поскольку признаки с большим масштабом могут доминировать над признаками с меньшим масштабом, что приводит к неоптимальной или нестабильной работе.

Примеры, когда нормализация мандаторна

  1. Алгоритмы, основанные на расстоянии (Distance-based Algorithms):
      - **k-ближайших соседей (k-NN) и k-средних (k-Means):** Эти алгоритмы используют Евклидово расстояние. Если один признак измеряется в миллионах (например, "Цена дома"), а другой — в десятках (например, "Количество комнат"), расстояние будет почти полностью определяться признаком с большим масштабом. Нормализация обеспечивает равный вклад всех признаков.

  2. Алгоритмы, основанные на градиентном спуске (Gradient Descent-based Algorithms):

      - **Линейная регрессия с SGD/Регуляризацией (Ridge, Lasso)**: В SGD ненормированные признаки приводят к сильно вытянутому (эллиптическому) ландшафту функции потерь. Градиентный спуск будет "метаться" туда-сюда, делая обучение очень медленным и нестабильным, так как для разных признаков потребуется разный шаг обучения (η).

  3. **Нейронные сети (Neural Networks)**: Нормализация входных данных (обычно в диапазон [0, 1] или со средним 0 и дисперсией 1) помогает стабилизировать активации нейронов и предотвращает проблему исчезающих/взрывающихся градиентов (vanishing/exploding gradients) в начальных слоях.


Примеры, когда нормализация необязательна/не нужна

  1. **Деревья решений (Decision Trees) и Случайный лес (Random Forest):**
      - Эти алгоритмы основаны на пороговых значениях и ранжировании (порядке) признаков, а не на расстоянии. Они выбирают точку разделения, которая максимизирует прирост информации или уменьшает энтропию. Масштаб признака не влияет на то, какое пороговое значение будет выбрано.

  2. **Наивный Байес (Naive Bayes):**
      - Этот алгоритм основан на вероятностях и не чувствителен к масштабу признаков, хотя его вариант для непрерывных данных (Гауссовский Наивный Байес) может работать лучше с нормированными данными.

  3. **Линейная регрессия с Аналитическим решением (OLS):**
      - Если вы используете метод Наименьших Квадратов (формула np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)), масштабирование не изменит конечных весов (хотя интерпретация весов изменится). Нормализация не нужна для получения корректного решения.

### 2. Let's consider the first of the classical normalization methods — MinMaxScaler. Write a mathematical formula for this method.

Xscaled = (X - Xmin) / (Xmax - Xmin)

### 3. Implement your own function or class for MinMaxScaler feature normalization.

```python
class MyMinMaxScaler:
  """
  Кастомный класс, реализующий Min-Max Feature Scaling.
  Масштабирует данные в диапазон [0, 1].
  """
  def __init__(self):
    # Хранит минимальные значения для каждого признака
    self.min_ = None
    # Хранит размах (max - min) для каждого признака
    self.range_ = None

  def fit(self, X):
    """
    Вычисляет минимальное значение (min) и размах (max - min) 
    для каждого столбца (признака) в тренировочном наборе X.
    
    Args:
        X (np.ndarray): Тренировочный набор данных.
    
    Returns:
        self: Возвращает сам объект.
    """
    # Если X_train имеет форму (m,), преобразуем его в (m, 1) для корректной работы
    if X.ndim == 1:
        X = X.reshape(-1, 1)

    # Вычисляем минимум по каждому столбцу (axis=0)
    self.min_ = np.min(X, axis=0)
    
    # Вычисляем максимум и размах
    max_ = np.max(X, axis=0)
    self.range_ = max_ - self.min_
    
    # Защита от деления на ноль: если range=0 (все значения одинаковы), 
    # range_ должен быть 1, чтобы избежать ошибки.
    self.range_[self.range_ == 0] = 1
    
    return self

  def transform(self, X):
    """
    Масштабирует данные X, используя сохраненные min и range.
    
    Args:
        X (np.ndarray): Набор данных для преобразования (Train или Test).
        
    Returns:
        np.ndarray: Масштабированный набор данных.
    """
    if self.min_ is None or self.range_ is None:
        raise Exception("First call fit() method")
        
    if X.ndim == 1:
        X = X.reshape(-1, 1)
    
    # Применяем формулу: (X - min) / range
    X_scaled = (X - self.min_) / self.range_

    return X_scaled

  def fit_transform(self, X):
    """
    Сначала вызывает fit(X), а затем transform(X).
    """
    self.fit(X)
    return self.transform(X)
```

### 6. Repeat the steps from b to e for another normalization method StandardScaler.

Формула : z = X - u / q, где u - среднее арифметическое, а q - стандартное отклонение

```python
class MyStandardScaler:
    """
    Кастомный класс, реализующий Standard Scaler (Z-score нормализацию).
    Центрирует данные (среднее = 0) и приводит дисперсию к 1.
    """
    def __init__(self):
        # Хранит среднее значение (mu) для каждого признака
        self.mean_ = None
        # Хранит стандартное отклонение (sigma) для каждого признака
        self.std_ = None
        
    def fit(self, X):
        """
        Вычисляет среднее значение (mean) и стандартное отклонение (std) 
        для каждого столбца (признака) в тренировочном наборе X.
        
        Args:
            X (np.ndarray): Тренировочный набор данных.
        
        Returns:
            self: Возвращает сам объект.
        """
        # Преобразуем 1D массив в 2D, если необходимо
        if X.ndim == 1:
            X = X.reshape(-1, 1)

        # Вычисляем среднее значение по каждому столбцу (axis=0)
        self.mean_ = np.mean(X, axis=0)
        
        # Вычисляем стандартное отклонение по каждому столбцу (axis=0)
        # В NumPy по умолчанию используется выборочное стандартное отклонение (ddof=0 для std)
        self.std_ = np.std(X, axis=0)
        
        # Защита от деления на ноль: если std=0 (все значения одинаковы), 
        # std_ должен быть 1, чтобы избежать ошибки.
        self.std_[self.std_ == 0] = 1
        
        return self

    def transform(self, X):
        """
        Масштабирует данные X, используя сохраненные mean и std.
        
        Args:
            X (np.ndarray): Набор данных для преобразования (Train или Test).
            
        Returns:
            np.ndarray: Стандартизированный набор данных.
        """
        if self.mean_ is None or self.std_ is None:
            raise Exception("Сначала необходимо вызвать метод fit()")
            
        if X.ndim == 1:
            X = X.reshape(-1, 1)
        
        # Применяем формулу: (X - mu) / sigma
        X_scaled = (X - self.mean_) / self.std_
        
        return X_scaled

    def fit_transform(self, X):
        """
        Сначала вызывает fit(X), а затем transform(X).
        """
        self.fit(X)
        return self.transform(X)
```

```python
def get_metrics_results(model, X_train, y_train, X_test, y_test, model_name):
    """Обучает модель, делает предсказания и возвращает словарь результатов."""
    
    # Обучение
    model.fit_sgd(X_train, y_train) 

    # Предсказания
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    # Сбор результатов
    results = [
        {'Model': model_name, 'Metric': 'R2', 'Train': r2_score(y_train, y_train_pred), 'Test': r2_score(y_test, y_test_pred)},
        {'Model': model_name, 'Metric': 'MAE', 'Train': mae_score(y_train, y_train_pred), 'Test': mae_score(y_test, y_test_pred)},
        {'Model': model_name, 'Metric': 'RMSE', 'Train': rmse_score(y_train, y_train_pred), 'Test': rmse_score(y_test, y_test_pred)}
    ]
    return results
```

```python
# 1. Определение моделей с фиксированными параметрами
models_to_test = [
    # Линейная регрессия (Без регуляризации)
    ('Poly LR (a=1.0)', LinearRegressionCustom(alpha=1.0, l1_ratio=0.0, n_epochs=100, learning_rate=0.01)),

    # Ridge (L2)
    ('Poly Ridge (a=1.0)', LinearRegressionCustom(alpha=1.0, l1_ratio=0.0, n_epochs=100, learning_rate=0.01)),
    
    # Lasso (L1)
    ('Poly Lasso (a=1.0)', LinearRegressionCustom(alpha=1.0, l1_ratio=1.0, n_epochs=100, learning_rate=0.01)),
    
    # ElasticNet
    ('Poly ElasticNet (a=1.0)', LinearRegressionCustom(alpha=1.0, l1_ratio=0.5, n_epochs=100, learning_rate=0.01)),
]

all_results_initial = []

# 2. Обучение и сбор метрик
for name, model in models_to_test:
    # Важно: для демонстрации мы используем низкий learning_rate и высокое n_epochs
    print(f"Обучение: {name}...")
    results = get_metrics_results(
        model, 
        X_poly_train_scaled, y_train, 
        X_poly_test_scaled, y_test, 
        name
    )

    all_results_initial.extend(results)

# 3. Создание DataFrame
df_initial_metrics = pd.DataFrame(all_results_initial)
```

### 1. Calculate the mean and median metrics from the previous lesson and add the results to the final dataframe.

```python
# 1. Вычисление наивных предсказаний
# Baseline 1: Среднее значение (mean) тренировочного набора
mean_y_train = np.mean(y_train)

# Предсказание для всех примеров Train и Test - это просто константа!
y_train_pred_mean = np.full(y_train.shape, mean_y_train)
y_test_pred_mean = np.full(y_test.shape, mean_y_train)

# Baseline 2: Медианное значение (median) тренировочного набора
median_y_train = np.median(y_train)
y_train_pred_median = np.full(y_train.shape, median_y_train)
y_test_pred_median = np.full(y_test.shape, median_y_train)

# 2. Функция для сбора метрик Baseline
def get_baseline_metrics(y_train, y_train_pred, y_test, y_test_pred, model_name):
    """Считает метрики для наивного предсказания и возвращает список результатов."""
    return [
        # R2
        {'Model': model_name, 'Metric': 'R2', 'Scale': 'None', 
         'Train': r2_score(y_train, y_train_pred), 'Test': r2_score(y_test, y_test_pred)},
        # MAE
        {'Model': model_name, 'Metric': 'MAE', 'Scale': 'None', 
         'Train': mae_score(y_train, y_train_pred), 'Test': mae_score(y_test, y_test_pred)},
        # RMSE
        {'Model': model_name, 'Metric': 'RMSE', 'Scale': 'None', 
         'Train': rmse_score(y_train, y_train_pred), 'Test': rmse_score(y_test, y_test_pred)}
    ]

# 3. Сбор результатов
baseline_results = []

# Baseline 1: Mean (Среднее)
baseline_results.extend(get_baseline_metrics(
    y_train, y_train_pred_mean, 
    y_test, y_test_pred_mean, 
    'Baseline_Mean'
))

# Baseline 2: Median (Медиана)
baseline_results.extend(get_baseline_metrics(
    y_train, y_train_pred_median, 
    y_test, y_test_pred_median, 
    'Baseline_Median'
))

# 4. Объединение с финальным DataFrame
df_baseline = pd.DataFrame(baseline_results)

# Добавляем наивные результаты к уже собранным результатам
all_results.extend(baseline_results)

# Создание итогового DataFrame (с пересборкой из all_results)
df_metrics_final = pd.DataFrame(all_results)
df_metrics_final = df_metrics_final.sort_values(by=['Model', 'Scale', 'Metric']).reset_index(drop=True)
```
